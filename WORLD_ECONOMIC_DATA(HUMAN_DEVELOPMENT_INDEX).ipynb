```python
QUERY_Q3="""
WITH StatComponents AS (
    SELECT
        COUNT(*) AS N,
        SUM(le_2021) AS Sum_X,
        SUM(gnipc_2021) AS Sum_Y,
        SUM(le_2021 * le_2021) AS Sum_X2,
        SUM(gnipc_2021 * gnipc_2021) AS Sum_Y2,
        SUM(le_2021 * gnipc_2021) AS Sum_XY
    FROM hdi_table
    WHERE le_2021 IS NOT NULL
      AND gnipc_2021 IS NOT NULL
)


SELECT
    --Covariance numerator: (N * ΣXY) - (ΣX * ΣY)
    ((N * Sum_XY) - (Sum_X * Sum_Y))
    /
    --Standard deviation denominator: √[N*ΣX² - (ΣX)²] * √[N*ΣY² - (ΣY)²]
    (
        SQRT((N * Sum_X2) - (Sum_X * Sum_X))
        *
        SQRT((N * Sum_Y2) - (Sum_Y * Sum_Y))
    ) AS pearson_correlation_r
FROM StatComponents;
"""
df_hdi_stats_query3=pd.read_sql_query(QUERY_Q3,conn)
df_hdi_stats_query3
```

```python
query_4 = """
SELECT
    -- 1. General Metrics
    COUNT(CASE WHEN hdicode = 'High' THEN 1 END) AS high_inc_count, -- Changed 'High Income' to 'High'
    COUNT(CASE WHEN hdicode = 'Low' THEN 1 END) AS low_inc_count, -- Changed 'Low Income' to 'Low'

    -- 2. Life Expectancy Comparisons & Gap
    ROUND(AVG(CASE WHEN hdicode = 'High' THEN le_2021 END), 1) AS high_inc_avg_life_exp, -- Changed 'High Income' to 'High', 'life_expectancy' to 'le_2021'
    ROUND(AVG(CASE WHEN hdicode = 'Low' THEN le_2021 END), 1) AS low_inc_avg_life_exp, -- Changed 'Low Income' to 'Low', 'life_expectancy' to 'le_2021'
    ROUND(
        AVG(CASE WHEN hdicode = 'High' THEN le_2021 END) -  -- Changed 'High Income' to 'High', 'life_expectancy' to 'le_2021'
        AVG(CASE WHEN hdicode = 'Low' THEN le_2021 END), 1 -- Changed 'Low Income' to 'Low', 'life_expectancy' to 'le_2021'
    ) AS life_exp_gap_years,

    -- 3. Education Comparisons & Gap
    ROUND(AVG(CASE WHEN hdicode = 'High' THEN mys_2021 END), 1) AS high_inc_avg_schooling, -- Changed 'High Income' to 'High', 'mean_years_schooling' to 'mys_2021'
    ROUND(AVG(CASE WHEN hdicode = 'Low' THEN mys_2021 END), 1) AS low_inc_avg_schooling, -- Changed 'Low Income' to 'Low', 'mean_years_schooling' to 'mys_2021'
    ROUND(
        AVG(CASE WHEN hdicode = 'High' THEN mys_2021 END) -  -- Changed 'High Income' to 'High', 'mean_years_schooling' to 'mys_2021'
        AVG(CASE WHEN hdicode = 'Low' THEN mys_2021 END), 1 -- Changed 'Low Income' to 'Low', 'mean_years_schooling' to 'mys_2021'
    ) AS schooling_gap_years,

    -- 4. Economic (GNI) Comparisons & Multiplier
    ROUND(AVG(CASE WHEN hdicode = 'High' THEN gnipc_2021 END), 0) AS high_inc_avg_gni, -- Changed 'High Income' to 'High', 'gni_per_capita' to 'gnipc_2021'
    ROUND(AVG(CASE WHEN hdicode = 'Low' THEN gnipc_2021 END), 0) AS low_inc_avg_gni, -- Changed 'Low Income' to 'Low', 'gni_per_capita' to 'gnipc_2021'
    ROUND(
        AVG(CASE WHEN hdicode = 'High' THEN gnipc_2021 END) /  -- Changed 'High Income' to 'High', 'gni_per_capita' to 'gnipc_2021'
        NULLIF(AVG(CASE WHEN hdicode = 'Low' THEN gnipc_2021 END), 0), 1 -- Changed 'Low Income' to 'Low', 'gni_per_capita' to 'gnipc_2021'
    ) AS gni_multiplier_fold

FROM hdi_table -- Changed 'country_metrics' to 'hdi_table'
WHERE hdicode IN ('High', 'Low'); -- Changed 'income_group' to 'hdicode' and 'High Income'/'Low Income' to 'High'/'Low'
"""
df_hdi_stats_query_4=pd.read_sql_query(query_4,conn)
df_hdi_stats_query_4
```

```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, round
from pyspark.sql.window import Window
from pyspark.sql.functions import desc

spark = SparkSession.builder \
    .appName("HDIGrowthOverlapAnalysis") \
    .getOrCreate()

df = spark.read.parquet("hdi.parquet")

growth_df = df.withColumn(
    "gni_growth_pct",
    round(((col("gnipc_2021") - col("gnipc_1990")) / col("gnipc_1990")) * 100, 2)
)



top_gni = growth_df.select("country", "gni_growth_pct") \
    .orderBy(desc("gni_growth_pct")) \
    .limit(20)



print("TOP 10 GNI GROWTH COUNTRIES")
top_gni.show(10, truncate=False)

```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, round, desc

spark = SparkSession.builder.appName("RegionalHDIGrowth").getOrCreate()


df = spark.read.parquet("hdi.parquet")

regional_growth = df.groupBy("region") \
    .agg(
        round(avg("hdi_2000"), 3).alias("avg_hdi_2000"),
        round(avg("hdi_2021"), 3).alias("avg_hdi_latest"),
        round(
            (avg("hdi_2021") - avg("hdi_2000")), 3
        ).alias("absolute_hdi_growth"),
        round(
            ((avg("hdi_2021") - avg("hdi_2000")) / avg("hdi_2000")) * 100, 1
        ).alias("percentage_growth_pct")
    ) \
    .orderBy(desc("absolute_hdi_growth"))

regional_growth.show(truncate=False)
```

```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, count, round


spark = SparkSession.builder \
    .appName("IncomeGroupDifferentiation") \
    .getOrCreate()
df = spark.read.parquet("hdi.parquet")

analysis_df = df \
    .filter(col("hdicode").isin("High", "Low")) \
    .groupBy("hdicode") \
    .agg(
        count("country").alias("total_countries"),
        round(avg("le_2021"), 1).alias("avg_life_exp"),
        round(avg("mys_2021"), 1).alias("avg_schooling_years"),
        round(avg("eys_2021"), 1).alias("avg_expected_schooling"),
        round(avg("gnipc_2021"), 0).alias("avg_gni_per_capita"),
        round(avg("hdi_2021"), 3).alias("avg_hdi")
    )


print("DECISIVE STRUCTURAL VARIANCES (HIGH VS LOW INCOME)")
analysis_df.show(truncate=False)


# spark.stop() # Removed spark.stop() to keep the session alive
```
